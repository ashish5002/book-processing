```python
import os
import fitz  # PyMuPDF
from PIL import Image
import io

def extract_pages_to_images(pdf_path, output_dir="extracted_pages", pages_per_case=10):
    """
    Extract pages from a PDF and save 10 individual JPG images per case folder.
    
    Args:
        pdf_path (str): Path to the input PDF file
        output_dir (str): Directory to save the extracted images
        pages_per_case (int): Number of pages to extract per case (default: 10)
    """
    
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    try:
        # Open the PDF document
        doc = fitz.open(pdf_path)
        total_pages = doc.page_count
        
        print(f"Processing PDF with {total_pages} pages...")
        print(f"Creating cases with {pages_per_case} images each...")
        
        case_number = 1
        extracted_count = 0
        
        # Process pages in chunks
        for start_page in range(0, total_pages, pages_per_case):
            end_page = min(start_page + pages_per_case, total_pages)
            
            # Create case directory
            case_dir = os.path.join(output_dir, f"case{case_number:04d}")
            os.makedirs(case_dir, exist_ok=True)
            
            # Extract individual pages for this case
            for page_num in range(start_page, end_page):
                page = doc[page_num]
                
                # Convert page to image (300 DPI for good quality)
                mat = fitz.Matrix(300/72, 300/72)  # 300 DPI scaling
                pix = page.get_pixmap(matrix=mat)
                img_data = pix.tobytes("ppm")
                img = Image.open(io.BytesIO(img_data))
                
                # Generate filename for individual page
                page_filename = f"page_{(page_num - start_page + 1):02d}.jpg"
                page_filepath = os.path.join(case_dir, page_filename)
                
                # Save as JPG
                img.save(page_filepath, "JPEG", quality=95)
            
            print(f"Saved: case{case_number:04d} with {end_page - start_page} images (pages {start_page + 1}-{end_page})")
            extracted_count += 1
            case_number += 1
        
        doc.close()
        print(f"\nExtraction complete! Created {extracted_count} case folders.")
        print(f"Images saved in: {output_dir}")
        
    except Exception as e:
        print(f"Error processing PDF: {str(e)}")

# Example usage
if __name__ == "__main__":
    # Replace with your PDF file path
    pdf_file = r'C:\Users\ashis\Desktop\book_processing\book.pdf'
    
    # Extract pages - creates 10 individual JPG images per case folder
    print("Extracting pages into case folders...")
    extract_pages_to_images(pdf_file, "static/extracted_cases", pages_per_case=10)
```
